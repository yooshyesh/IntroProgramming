# Klassen und Instanzen
Wenn wir ein Programm schreiben, haben wir es häufig mit Objekten zu tun, die eine ähnliche Struktur haben. In einer Bibliothek können wir verschiedene Medien ausleihen. Dies können Bücher, DVDs oder eBooks sein. Diese Medien haben Gemeinsamkeiten und Unterschiede. Für die Verwaltung in der Bibliothek ist es notwendig, dass alle Medien über eine eindeutige Identifikationsnummer und einen Titel verfügen, damit diese katalogisiert werden können.

Die Struktur fassen wir in einer *Klasse* zusammen. Die Klasse ```Media``` beschreibt, was *alle* Medien gemeinsam haben. Alle Medien haben eine ```id``` und einen ```title```. Diese beiden Felder werden als Attribute der Klasse bezeichnet.

## Der Konstruktor
Die Methode ```def __init__(self, id, title):``` ist eine besondere Methode. Diese Methode wird immer dann aufgerufen, wenn eine *Instanz* der Klasse Media, also zum Beispiel ein Buch, erzeugt wird. Die Methode wird häufig auch als Konstruktor bezeichnet. Wegen der zwei Unterstriche, die die Methode umgeben, heissen solche Methoden auch dunder-Methoden (*double underscore*).

## Instanzen
In der letzten Zeile des Codebeispiels wird eine Instanz mit dem Namen ```buch1``` erzeugt. Eine Instanz ist eine konkretes Exemplar einer Klasse. Ein Objekt ist eine Instanz einer Klasse. Eine Klasse ist wie ein Bauplan für Objekte. Die Startwerte für die beiden Attribute werden als Prameter an den Konstruktor übergeben.

````python
class Media:
    def __init__(self, id, title):
        self.id = id
        self.title = title

buch1 = Media(1, "Der Prozess")
print(buch1.title)
````

Es gibt aber auch Unterschiede zwischen den Medien. So verfügen Bücher über eine Seitenzahl und einen Autor.



````python
class Media:
    def __init__(self, id, title):
        self.id = id
        self.title = title

class Book(Media):
    def __init__(self, id, title, pages, author):
        super().__init__(id, title)
        self.pages = pages
        self.author = author

mein_buch = Book(1, "Der Prozess", 240, "Franz Kafka")
print(f"{mein_buch.title} von {mein_buch.author}, {mein_buch.pages} Seiten")
````
