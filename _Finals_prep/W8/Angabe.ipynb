Zähle, wie viele Kommentare positiv, negativ, oder neutral sind. Nutze dafür Schleifen und Bedingungen.

````python
kommentare = [
    ("Die Navigation war klar.", "positiv"),
    ("Ich habe den Warenkorb nicht gefunden.", "negativ"),
    ("Zu viele Klicks nötig.", "negativ"),
    ("Sehr gute Suchfunktion!", "positiv")
]

positiv:int = 0
negativ:int = 0
neutral:int = 0

for bewertung in kommentare:
    if bewertung[1] == "positiv":
        positiv += 1
    elif bewertung[1] == "negativ":
        negativ += 1
    elif bewertung[1] == "neutral":
        neutral += 1

print(f"Positive Kommentare: {positiv}")
print(f"Negative Kommentare: {negativ}")
print(f"Neutrale Kommentare: {neutral}")
````

Der Umgang mit Tuples ist nicht sehr praktisch. Es ist einfacher, auf die Felder über einen Namen zurückzugreifen. Tausche die Implementierung nun durch ein Dictionary (Schlüssel "text" und "bewertung") aus.

````python
kommentare = [
    {"text": "Die Navigation war klar.", "bewertung": "positiv"},
    {"text": "Ich habe den Warenkorb nicht gefunden.", "bewertung": "negativ"},
    {"text": "Zu viele Klicks nötig.", "bewertung": "negativ"},
    {"text": "Sehr gute Suchfunktion!", "bewertung": "positiv"}
]

positiv:int = 0
negativ:int = 0
neutral:int = 0

for rating in kommentare:
    if rating["bewertung"] == "positiv":
        positiv += 1
    elif rating["bewertung"] == "negativ":
        negativ += 1
    elif rating["bewertung"] == "neutral":
        neutral += 1

print(f"Positive Kommentare: {positiv}")
print(f"Negative Kommentare: {negativ}")
print(f"Neutrale Kommentare: {neutral}")
````

Schreibe eine Funktion, die das Dictionary `kommentare` als Parameter erhält und in ein Dictionary überführt, dass nach Bewertung gruppiert ist. Beachte die Beispielausgabe.

````python
def gruppiere_nach_bewertung(kommentare):
    gruppiert = {} # Äussere Hülle = leeres dictionary
    for entry in kommentare:
        bewertung = entry["bewertung"] # variable von keys erstellen
        text = entry["text"]
        if bewertung not in gruppiert: # falls noch nicht in gruppiert
            gruppiert[bewertung] = [] # Innere Hülle = leere Liste, bewertung als dict key setzen
        gruppiert[bewertung].append(text) # text in Liste einfügen

kommentare = [
    {"text": "Die Navigation war klar.", "bewertung": "positiv"},
    {"text": "Ich habe den Warenkorb nicht gefunden.", "bewertung": "negativ"},
    {"text": "Zu viele Klicks nötig.", "bewertung": "negativ"},
    {"text": "Sehr gute Suchfunktion!", "bewertung": "positiv"}
]

gruppiert = gruppiere_nach_bewertung(kommentare)
print(gruppiert)

# {
#     'positiv': ['Die Navigation war klar.', 'Sehr gute Suchfunktion!'],
#     'negativ': ['Ich habe den Warenkorb nicht gefunden.', 'Zu viele Klicks nötig.']
# }
````
