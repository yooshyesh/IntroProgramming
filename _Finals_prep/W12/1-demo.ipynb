```python
class Auto:
    # Klassenattribut (gilt für alle Instanzen)
    anzahl_raeder = 4

    def __init__(self, farbe):
        # Instanzattribut (individuell pro Objekt)
        self.farbe = farbe

# Zwei Instanzen der Klasse Auto
auto1 = Auto("rot")
auto2 = Auto("blau")

# Zugriff auf Instanzattribute
print("Auto1 Farbe:", auto1.farbe)  # rot
print("Auto2 Farbe:", auto2.farbe)  # blau

# Zugriff auf Klassenattribut
print("Auto1 Räder:", auto1.anzahl_raeder)  # 4
print("Auto2 Räder:", auto2.anzahl_raeder)  # 4

# Änderung des Klassenattributs
Auto.anzahl_raeder = 3
print("Auto1 Räder nach Änderung:", auto1.anzahl_raeder)  # 3
print("Auto2 Räder nach Änderung:", auto2.anzahl_raeder)  # 3

# Änderung eines Instanzattributs
auto1.farbe = "grün"
print("Auto1 neue Farbe:", auto1.farbe)  # grün
print("Auto2 Farbe bleibt:", auto2.farbe)  # blau
```
