## Klassen, Attribute und Methoden
- Eine Klasse ist ein Bauplan für Objekte.
  - Klasse = Bauplan
  - Instanz = konkretes Objekt von einer Klasse
- Attribute beschreiben die **Eigenschaften**.
- Methoden beschreiben das **Verhalten**.

````python
class Person:
    # Konstruktor
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter

    # Methode
    def vorstellen(self):
        print(f"Hallo, ich bin {self.name}.")

# Erzeugen und Verwendung einer Instanz
frieder = Person("Frieder", "12")
frieder.vorstellen()

# Es kann auch direkt auf Attribute einer Instanz zugegriffen werden
print(frieder.name)
print(frieder.alter)
````

## Dunder-Methoden
- `__init__`: Konstruktor
- `__str__`: lesbare Darstellung für `print()`

Diese Methoden werden wegen der beiden Unterstriche (**d**ouble **under**score) als Dunder-Methoden bezeichnet. Es gibt noch weitere Dunder-Methoden.

````python
class Person:
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter
    
    def __str__(self):
        return f"{self.name}, {self.alter} Jahre alt"

print(Person("Alice", 30)) # Die Ausgabe von print wird mit der Methode __str__ erzeugt.
````

## Getter und Setter
- Schützen Attribute vor ungewünschten Werten
- Ermöglichen direkten Zugriff

````python
class Person:
    def __init__(self, name, alter = 0): # 0 Standardwert
        self.name = name
        self.alter = alter 
	
	# getter Methode
    def get_alter(self): 
        return self.alter 
	
	# setter Methode
    def set_alter(self, x): 
        self.alter = x 

studentin = Person("Moira", 19) 

print(studentin.get_alter()) 

studentin.set_alter(20) # Moira's bday was last week

# Alternative Scchreibweise
print(studentin.alter)
````

### Aufgabe
Erweitere die `Auto`-Klasse mit Getter und Setter für das Baujahr. Stelle in deiner setter-Methode sicher, dass kein `baujahr` kleiner als 1886 (davor gab es noch keine Autos) gesetzt werden darf.

````python
# Todo...
class Auto:
    def __init__(self, marke, baujahr):
        self.marke = marke
        self.baujahr = baujahr

    def fahren(self):
        print(f"{self.marke} fährt los.") # wird mit self. wie oben definiert

    def __str__(self):
        return f"{self.marke}, {self.baujahr} gebaut."
    
    def get_baujahr(self):
        return self.baujahr
    
    def set_baujahr(self, x): # self muss angegeben werden, damit betroffenes Objekt definiert wird
        if x < 1886:
            raise ValueError("Ungültiges Baujahr")
        self.baujahr = x # baujahr erst nach restrictions definieren lassen

Auto1 = Auto("VW", 2002)
Auto2 = Auto("Miata", "2005")
Auto3 = Auto("Audi", "2022")

print(Auto1.get_baujahr())

Auto1.set_baujahr(2005)
print(Auto1.baujahr)
Auto2.set_baujahr(1700)
print(Auto2.baujahr)
````

## Datenstrukturen in Klassen
- Eine Klasse kann Listen, Dictionaries etc. als Attribute enthalten
- Beispiel: Sammlung von Objekten verwalten

````python
class Klasse:
    def __init__(self):
        self.schueler = []
    
    def hinzufuegen(self, person):
        self.schueler.append(person)
    
    def anzeigen(self):
        for p in self.schueler:
            print(p)
````

### Aufgabe
Schreibe eine Klasse `Garage`, die mehrere `Auto`-Objekte speichern kann. Sie soll Methoden haben, um Autos hinzuzufügen und alle Autos auszugeben.

````python
class Auto:
    def __init__(self, marke, baujahr):
        self.marke = marke
        self.baujahr = baujahr

    def __str__(self):
        return f"{self.marke} ({self.baujahr})"

class Garage(Auto):
    def __init__(self):
        self.autos = []

    def add(self, Auto):
        self.autos.append(Auto)

    def show(self):
        for a in self.autos:
            print(a)


# Test
g = Garage()
g.add(Auto("BMW", 2019))
g.add(Auto("Audi", 2021))
g.show()

# Erwartete Ausgabe
# BMW (2019)
# Audi (2021)
````
