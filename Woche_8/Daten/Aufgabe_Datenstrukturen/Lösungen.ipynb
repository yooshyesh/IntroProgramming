## Liste (List) – Quizauswertung

```python
scores: list[int] = [58, 72, 0, 91, 65]

total = 0
count = 0
for score in scores:
    if score != 0: # 0 wird ignoriert
        total += score
        count += 1

avg = total / count if count else 0
if avg >= 60:
    print("Kursziel erreicht:", round(avg, 1))
else:
    print("Kursziel nicht erreicht:", round(avg, 1))

scores.append(77)
print("Aktualisierte Liste:", scores)
```

## Tupel (Tuple) – Temperaturbereich überwachen

```python
allowed: tuple[float, float] = (18.0, 24.0)
temps: list[float] = [19.5, 25.0, 22.1, 17.9, 21.3]

ok = 0
for t in temps:
    if t < allowed[0] or t > allowed[1]:
        print(f"Warnung: {t}°C ausserhalb von {allowed}")
    else:
        ok += 1

ratio = (ok / len(temps)) * 100
print(f"Innerhalb des Bereichs: {ratio:.1f}%")
```

## Sets

```python
registered: set[str] = {"Anna", "Bob", "Clara", "Diego"}
checkins: list[str] = ["Anna", "Edo", "Clara", "Anna"]

present, unknown = set(), set()

for name in checkins:
    if name in registered:
        present.add(name)
    else:
        unknown.add(name)

quote = (len(present) / len(registered)) * 100
print(f"Anwesenheitsquote: {quote:.1f}%")
print("Anwesend:", present)
print("Unbekannt:", unknown)
```

## Dictionary – Notenbuch

```python
grades: dict[str, float] = {"Anna": 5.0, "Bob": 4.0, "Clara": 5.5, "Diego": 3.5}

names = list(grades.keys())
i = 0
total = 0

for name in names:
    note = grades[name]
    status = "bestanden" if note >= 4.0 else "nicht bestanden"
    print(f"{name}: {note:.1f} – {status}")
    total += note

avg = total / len(grades) if grades else 0
print(f"Klassendurchschnitt: {avg:.2f}")
```
