## Aufgabe 1: Einstieg in Polymorphie
Erstelle eine Basisklasse Tier mit einer Methode `geraeusch()`.
Implementiere Unterklassen Löwe, Affe und Papagei, die unterschiedliche Geräusche zurückgeben.
Tier: `tierisch` / Affe: `UH UH AH AH!` / Loewe: `ROAR!` / Papagei: `Halooo!`

```python
# Musterlösung
class Tier:
    def geraeusch(self):
        return "tierisch"

class Loewe(Tier):
    def geraeusch(self):
        return "ROAR!"

class Affe(Tier):
    def geraeusch(self):
        return "UH UH AH AH!"

class Papagei(Tier):
    def geraeusch(self):
        return "Hallooo!"

tiere = [Loewe(), Affe(), Papagei()]
for t in tiere:
    print(t.geraeusch())
```

## Aufgabe 2: Polymorphie im Zoo-Alltag
Füttere verschiedene Tiere. Jeder bekommt anderes Futter.
`Der Elefant bekommt Früchte.` `Der Panda knabbert Bambus.` `Der Löwe frisst Fleisch.`

```python
# Musterlösung
class Tier:
    def fuettern(self):
        return "Futter..."

class Loewe(Tier):
    def fuettern(self):
        return "Der Löwe frisst Fleisch."

class Panda(Tier):
    def fuettern(self):
        return "Der Panda knabbert Bambus."

class Elefant(Tier):
    def fuettern(self):
        return "Der Elefant bekommt Früchte."

def fuetterungsrunde(tiere):
    for t in tiere:
        print(t.fuettern())

tiere = [Loewe(), Panda(), Elefant()]
fuetterungsrunde(tiere)
```

Polymorphie bedeutet in der objektorientierten Programmierung, dass Objekte verschiedener Klassen über eine gemeinsame Schnittstelle (hier: die Methode `geraeusch()` oder `fuettern()`) angesprochen werden, aber unterschiedlich reagieren. Das heisst: die gleiche Methode wird aufgerufen, aber das Verhalten hängt von der konkreten Unterklasse ab.

## Aufgabe 3: Duck Typing
Im Zoo gibt es Geräte, die alle eine Methode `auffuellen()` haben. Entweder wird `Futterautomat wird mit {menge} kg Futter gefüllt.` oder `Wasserquelle wird mit {menge} Litern Wasser gefüllt.`.

```python
# Musterlösung
class Futterautomat:
    def auffuellen(self, menge):
        print(f"Futterautomat wird mit {menge} kg Futter gefüllt.")

class Wasserquelle:
    def auffuellen(self, menge):
        print(f"Wasserquelle wird mit {menge} Litern Wasser gefüllt.")

def auffuellen_im_zoo(geraet):
    geraet.auffuellen(10)

f = Futterautomat()
w = Wasserquelle()
auffuellen_im_zoo(f)
auffuellen_im_zoo(w)
```

## Aufgabe 4: Interfaces mit `abc`-Modul
Definiere eine abstrakte Klasse TierTransport.

```python
# Musterlösung
from abc import ABC, abstractmethod

class TierTransport(ABC):
    @abstractmethod
    def transportiere(self, tier):
        pass

class KaefigTransport(TierTransport):
    def transportiere(self, tier):
        print(f"{tier} wird im Käfig transportiert.")

class BoxTransport(TierTransport):
    def transportiere(self, tier):
        print(f"{tier} wird in einer Box transportiert.")

def bring_zum_gehege(transport, tier):
    transport.transportiere(tier)

bring_zum_gehege(KaefigTransport(), "Löwe")
bring_zum_gehege(BoxTransport(), "Papagei")
```

## Aufgabe 5: Praxisprojekt: Zoo-Gehege
Erstelle ein Interface Gehege mit `anzahl_tiere()` und `flaeche()`.

Implementiere die Klassen:
`Aquarium`: Enthält 20 Tiere (z. B. Fische) auf einer Fläche von 200 m²
`Vogelvoliere`: Enthält 15 Tiere (z. B. Vögel) auf einer Fläche von 150 m²
`SavannenGehege`: Enthält 10 Tiere (z. B. Giraffen, Zebras, Löwen) auf einer Fläche von 500 m²

Schreibe ausserdem eine Funktion gesamtflaeche(gehege_liste), die die Fläche aller Gehege summiert, und eine Funktion zoo_report(gehege_liste), die für jedes Gehege Name, Tieranzahl und Fläche ausgibt.

```python
# Musterlösung
from abc import ABC, abstractmethod

class Gehege(ABC):
    @abstractmethod
    def anzahl_tiere(self):
        pass

    @abstractmethod
    def flaeche(self):
        pass

class Aquarium(Gehege):
    def anzahl_tiere(self):
        return 20
    def flaeche(self):
        return 200

class Vogelvoliere(Gehege):
    def anzahl_tiere(self):
        return 15
    def flaeche(self):
        return 150

class SavannenGehege(Gehege):
    def anzahl_tiere(self):
        return 10
    def flaeche(self):
        return 500

def gesamtflaeche(gehege_liste):
    return sum(g.flaeche() for g in gehege_liste)

# Test
def zoo_report(gehege_liste):
    for g in gehege_liste:
        print(f"{g.__class__.__name__}: {g.anzahl_tiere()} Tiere, Fläche {g.flaeche()} m²")
    print("Gesamtfläche:", gesamtflaeche(gehege_liste))

gehege = [Aquarium(), Vogelvoliere(), SavannenGehege()]
zoo_report(gehege)
```
