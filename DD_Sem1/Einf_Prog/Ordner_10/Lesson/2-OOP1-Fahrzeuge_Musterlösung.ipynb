## Klassen, Attribute und Methoden
- Eine Klasse ist ein Bauplan für Objekte.
- Attribute = Eigenschaften
- Methoden = Verhalten

````python
class Person:
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter
    
    def vorstellen(self):
        print(f"Hallo, ich heisse {self.name}.")
````

## Dunder-Methoden
- `__init__`: Konstruktor
- `__str__`: lesbare Darstellung für `print()`

````python
class Person:
    def __init__(self, name, alter):
        self.name = name
        self.alter = alter
    
    def __str__(self):
        return f"{self.name}, {self.alter} Jahre alt"

print(Person("Alice", 30))
````

### Aufgabe
Erweitere deine `Auto`-Klasse um eine `__str__`-Methode, sodass `print(auto)` eine schöne Ausgabe liefert. 

````python
# Musterlösung
class Auto:
    def __init__(self, marke, baujahr):
        self.marke = marke
        self.baujahr = baujahr
    
    def __str__(self):
        return f"{self.marke} (Baujahr {self.baujahr})"

# Test
print(Auto("VW Polo", 2022))
````

## Getter und Setter
- Schützen Attribute vor ungewünschten Werten
- Ermöglichen direkten Zugriff

````python
class Person:
    def __init__(self, name, alter = 0): 
        self.name = name
        self.alter = alter 
	
	# getter Methode
    def get_alter(self): 
        return self.alter 
	
	# setter Methode
    def set_alter(self, x): 
        self.alter = x 

studentin = Person("Moira", 19) 

print(studentin.get_alter()) 

studentin.set_alter(20) 

# Alternative Scchreibweise
print(studentin.alter)
````

### Aufgabe
Erweitere die `Auto`-Klasse mit Getter und Setter für das Baujahr. Stelle sicher, dass kein Jahr kleiner als 1886 (zuvor gab es noch keine Autos) gesetzt werden darf.

````python
# Musterlösung
class Auto:
    def __init__(self, marke, baujahr):
        self.marke = marke
        self.baujahr = baujahr

    def get_baujahr(self):
        return self.baujahr

    def set_baujahr(self, value):
        if value < 1886:
            raise ValueError("Baujahr ungültig! (erstes Auto 1886)") # oder Print + korrektes setzen
        self.baujahr = value

# Test
a = Auto("VW", 2015)
print(a.get_baujahr())   # Getter
a.set_baujahr(2020)   # Setter
print(a.baujahr)
````

## Datenstrukturen in Klassen
- Eine Klasse kann Listen, Dictionaries etc. enthalten
- Beispiel: Sammlung von Objekten verwalten

````python
class Klasse:
    def __init__(self):
        self.schueler = []
    
    def hinzufuegen(self, person):
        self.schueler.append(person)
    
    def anzeigen(self):
        for p in self.schueler:
            print(p)
````

## Vererbung

````python
class Tier:
    def __init__(self, name):
        self.name = name
    
    def geraeusch(self):
        print("Ein Tier macht ein Geräusch.")

class Hund(Tier):
    def geraeusch(self):
        print("Wuff!")
````

### Aufgabe
Erstelle eine Basisklasse `Fahrzeug` mit Attribut `marke` und Methode `fahren()`. Leite `Auto` und `Motorrad` davon ab. Überschreibe in `Motorrad` die Methode `fahren()`, sodass `"Motorrad <marke> fährt mit <hubraum>cc"` ausgegeben wird. `Auto` braucht das Attribut `baujahr`.

````python
# Dein Code hier

# Test
a = Auto("VW", 2018)
m = Motorrad("Honda", 1000)

a.fahren()
m.fahren()
print(a.baujahr)

# Erwartete Ausgabe:
# VW fährt los!
# Motorrad Honda fährt mit 1000cc
# 2018
````

````python
# Musterlösung
class Fahrzeug:
    def __init__(self, marke):
        self.marke = marke
    
    def fahren(self):
        print(f"{self.marke} fährt los!")

class Auto(Fahrzeug):
    def __init__(self, marke, baujahr):
        super().__init__(marke)
        self.baujahr = baujahr

class Motorrad(Fahrzeug):
    def __init__(self, marke, hubraum):
        super().__init__(marke)
        self.hubraum = hubraum
    
    def fahren(self):  # Methode überschrieben
        print(f"Motorrad {self.marke} fährt mit {self.hubraum}cc")

# Test
a = Auto("VW", 2018)
m = Motorrad("Honda", 1000)

a.fahren()
m.fahren()
print(a.baujahr)
````
