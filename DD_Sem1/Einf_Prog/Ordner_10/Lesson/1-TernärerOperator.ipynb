# Ternärer Operator

Wir möchten bestimmen, ob eine Zahl gerade oder ungerade ist. Hierfür ist eine Fallunterscheidung notwendig.

```python
n = 5

if n % 2 == 0:
    typ = "gerade"
else:
    typ = "ungerade"

print(typ)
```

Code dieser Art können wir mit einem _ternären Operator_ vereinfachen. Der folgende Code ist gleichwertig.

```python
n = 5
typ = "gerade" if n % 2 == 0 else "ungerade"
print(typ)
```

Der ternäre Operator hat die folgenden Bestandteile.

- **Bedingung:** `n % 2 == 0` Prüft, ob die Zahl n durch 2 teilbar ist (also gerade).
- **Wert, wenn Bedingung `True`:** "gerade" Wird zurückgegeben, wenn die Bedingung erfüllt ist.
- **Wert, wenn Bedingung `False`:** "ungerade" Wird zurückgegeben, wenn die Bedingung nicht erfüllt ist.
