# Liste Comprehension

List Comprehensions sind eine kompakte Methode, um Listen zu erstellen. Häufige Anwendungsfälle sind das Erzeugen neuer Listen, bei denen jedes Element das Ergebnis einer Operation ist, die auf jedes Element einer anderen Sequenz oder eines Iterables angewendet wird, oder das Erstellen einer Teilmenge jener Elemente, die eine bestimmte Bedingung erfüllen.

## Anwendungen

### Filtern von Elementen
Dies erleichtert das Filtern von Elementen aus einer Liste mit einer Bedingung.

```python
numbers = [-3, -1, 0, 2, 4]
# positives = [entry for entry in numbers if entry > 0] # List Comprehension
positives = [entry + 1 for entry in numbers if entry > 0] # List Comprehension
print(positives)

# simple Darstellung
positives = []
for n in numbers:
    if (n > 0):
        positives.append(n + 1)
```

Art der Werteinfügung: entry for entry/n + 1 for n
Obligatorischer Iterationsteil: entry in numbers
Optionale Bedingung: if entry > 0

Die List Comprehension hat die folgenden Bestandteile.
- `n` Jedes Element der Liste wird unverändert in die neue Liste eingefügt.
- `n in numbers` **Iterationsteil**, wir verwenden alle Teile der Liste `numbers`.
- `if n > 0` **Bedingungsteil**, Es werden nur Elemente verwendet, die diese Bedingung erfüllen.

### Quadrieren von allen Elementen einer Liste

```python
# Ohne List Comprehension
squares = []
for x in range(5):
    squares.append(x**2)

# Mit List Comprehension (x ** 2 -> x = x * x)
squares = [x**2 for x in range(5)]

print(squares)
```

Die List Comprehension hat die folgenden Bestandteile.
- `x**2` Jedes Element der Liste wird mit sich selbst multipliziert in die neue Liste eingefügt.
- `x in range(5)` **Iterationsteil**, `x` nimmt die Werte von 0 bis 4 an.

### Filtern von geraden Zahlen

```python
numbers = range(10)
evens = [n for n in numbers if n % 2 == 0]

print(evens)
```

### Zahlen kategorisieren

Mit Liste Comprehensions können auch verschachtelte Datenstrukturen erzeugt werden. Im folgenden Beispiel wird zu jeder Zahl hinzugefügt, ob sie gerade oder ungerade ist. Dies wird als Liste von Tupeln gespeichert.

```python
numbers = range(1, 11)
categories = [(n, "gerade" if n % 2 == 0 else "ungerade") for n in numbers]
print(categories)
```
