## Einführung in die Objektorientierte Programmierung mit Python

Als weiteres Beispiel für objektorientierte Programmierung, bauen wir nun eine eigene Datenstruktur. Die Klasse `SortedList` repräsentiert eine Liste, die ihre Elemente automatisch sortiert hält. Jedes Mal, wenn ein neues Element hinzugefügt wird, wird es an der richtigen Stelle eingefügt, um die Sortierreihenfolge zu bewahren.

Die Klasse definieren wir mit dem Keyword *class*, gefolgt von der Bezeichnung und einem Doppelpunkt.

```python
class SortedList:
```

Dann definieren wir als einen Konstruktor. Der Konstruktor wird für die Instanzierung verwendet, also wenn eine Instanz einer *SortedList* erzeugt wird. Dafür verwenden wir eine Funktion mit dem Namen `__init__`. Als Argument wird in Klassen `self` mitgegeben, also eine Referenz auf sich selbst. Damit erstellen wir eine leere Liste namens *elements*:

```python
class SortedList:
    def __init__(self):
        self.elements = []
```

### Attribute und Methoden der `SortedList`

- **Attribute**: Das wichtigste Attribut der Klasse `SortedList` ist eine Liste der gespeicherten Elemente. Dieses Liste wird im Konstrukor erstellt.
  
- **Methoden**:
  - `__init__`:  (Konstruktor) Initialisiert eine neue, leere Liste.
  - `add`: Fügt ein neues Element in die sortierte Liste ein. Dabei muss die Sortierreihenfolge erhalten bleiben.
  - `show_list`: Gibt die aktuelle Liste aus.

### Vollständiger Code

````python
class SortedList:
    def __init__(self):
        self.elements = []

    def add(self, element):
        # Insert element at the correct position to maintain sorting
        index = 0
        for current_element in self.elements:
            if current_element > element:
                break
            index += 1 # nicht if else, wird so oder so ausgeführt
        self.elements.insert(index, element)

    def show_list(self):
        print(self.elements)

# Using the SortedList class
my_list = SortedList()
my_list.add(5)
my_list.add(3)
my_list.add(8)

my_list.show_list()  # Outputs [3, 5, 8]
````

Die Klasse *SortedList* ist in dieser einfachen Implementation natürlich nur begrenzt nützlich. Man könnte noch diverse weitere Methoden ergänzen, wie zum Beispiel einzelne Elemente entfernen, die Position eines Elements zurückgeben oder die Liste in Listenform zur weiteren Verarbeitung zurückgeben. Dann könnte man auch die Liste so erweitern, dass man verschiedene Sortiermethoden verwenden kann, wie zum Beispiel auf- und absteigend.
