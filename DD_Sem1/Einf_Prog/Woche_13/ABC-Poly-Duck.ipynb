Abstract methods

```python
from abc import ABC, abstractmethod

class Tier(ABC): # jedes Tier hat eine Methode xy, durch ABC > Tierobjekt wird nie erzeugt
    @abstractmethod # Abstrakte Basisklasse, Methodenvererbung @ > decorator
    def geräusch(self):
        pass # Methode nicht in ABC definieren

class Hund(Tier):
    def geräusch(self):
        return "Wau"

class Katze(Tier):
    def geräusch(self):
        return "Miau"

# Gibt einen Fehler, da Tier eine abstrakte Klasse ist und nicht instanziiert werden kann.
# t = Tier()
# print(t.geräusch())

def geräusche(tiere):
    for t in tiere:
        print(t.geräusch())

tiere = [Hund(), Katze(), Katze()]
geräusche(tiere)
```

Polymorphism

Instanzen verschiedener Klassen werden über eine gemeinsame Schnittstelle (hier: die Methode geraeusch() oder fuettern()) angesprochen werden, aber unterschiedlich reagieren. Die gleiche Methode wird aufgerufen, aber das Verhalten hängt von der konkreten Unterklasse ab.

```python
class Tier:
    def fuettern(self):
        return "Futter..."

class Loewe(Tier):
    def fuettern(self):
        return "Der Löwe frisst Fleisch."

class Panda(Tier):
    def fuettern(self):
        return "Der Panda knabbert Bambus."

class Elefant(Tier):
    def fuettern(self):
        return "Der Elefant bekommt Früchte."

def fuetterungsrunde(tiere):
    for t in tiere:
        print(t.fuettern())

tiere = [Loewe(), Panda(), Elefant()]
fuetterungsrunde(tiere)
```
