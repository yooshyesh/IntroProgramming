Für häufige Aufgaben, kannst du Bibliotheken von Python verwenden.

1. **Random:** Die `random`-Bibliothek bietet Funktionen zur Erzeugung von Zufallszahlen. Damit kannst du zufällige Elemente aus einer Liste auswählen oder zufällige Daten für Tests erzeugen.

2. **Datetime:** Die `datetime`-Bibliothek unterstützt die Arbeit mit Datum und Uhrzeit. Sie ermöglicht die Ermittlung der aktuellen Zeit, die Berechnung von Zeitspannen und die Konvertierung von Datumsformaten.

3. **Math:** Die `math`-Bibliothek enthält mathematische Funktionen und Konstanten.

4. **IO:** Die `IO`-Bibliothek unterstützt das Lesen und Schreiben von Dateien. Sie wird in einem anderen Dokument beschrieben.

## Verwenden von Bibliotheken

Um eine Bibliothek zu verwenden, muss sie mit dem Befehl `import` eingebunden werden. In Jupyter-Notebooks musst du zuerst die Zelle ausführen, in der die Bibliothek eingebunden wird.

### Random

```python
import random

users = ["u001", "u002", "u003", "u004", "u005"]
selected = random.choice(users) # Zufälliges Element aus der Liste users auswählen

print(f"Zufällig ausgewählter User: {selected}")
```

```python
success = random.random() < 0.8  # success wird mit 80% Wahrscheinlichkeit auf True gesetzt
print("Aufgabe erfolgreich abgeschlossen:", success)
```

### Datetime

```python
import datetime

jetzt = datetime.datetime.now()
print("Aktuelles Datum und Uhrzeit:", jetzt)
```

```python
import datetime
import time

start = datetime.datetime.now() # Aktuelles Datum und Uhrzeit speichern
time.sleep(3)  # Simuliere eine Aufgabe, die 3 Sekunden dauert
end = datetime.datetime.now()

dauer = end - start
print(f"Nutzungsdauer: {dauer.total_seconds()} Sekunden")
```

### Math

```python
import math

# Arithmetisches Mittel berechnen
zeiten = [45, 60, 75, 90]
mittelwert = sum(zeiten) / len(zeiten)

print(f"Mittelwert: {mittelwert:.1f} s")
```

```python
# Runden von Werten
wert = 4.7
print("Gerundet:", math.round(wert))
print("Aufgerundet:", math.ceil(wert))
print("Abgerundet:", math.floor(wert))
```
