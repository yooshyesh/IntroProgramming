# Daten Lesen und Schreiben

Das Lesen und Schreiben von Dateien ist ein wichtiger Bestandteil vieler Programme. Im folgenden Beispiel wird eine Datei mit Namen `beispiel.txt` zunächst Geschrieben und dann Gelesen.

```python
# Datei schreiben
with open('beispiel.txt', 'w') as file:
    file.write("Hallo Welt!")

# Datei lesen
with open('beispiel.txt', 'r') as file:
    inhalt = file.read()
    print(inhalt)
```

## Öffnen einer Datei

Wir können eine Datei mit der Funktion `open` öffnen. Diese Funktion erwartet mindestens einen Parameter: den Dateipfad. Optional kann ein Modus angegeben werden (z.B. `'r'` für Lesen oder `'w'` für Schreiben).

`with open(...)` sorgt dafür, dass die Datei nach dem Lesen oder Schreiben automatisch geschlossen wird. Das ist wichtig, denn offene Dateien können Ressourcen binden, wenn sie am Ende nicht geschlossen werden.

## Lesen einer Datei

Zum Lesen einer Datei öffnest du sie im Lese-Modus (`'r'`). Du kannst entweder die gesamte Datei auf einmal lesen (`read()`) oder zeilenweise vorgehen (`readline()` oder `readlines()`).

Achte bei den folgenden Beispielen darauf, dass die Dateien `beispiel.txt`, `sessions.csv` und `sessions.json` im selben Ordner wie diese Datei liegen.

### JSON-Datei lesen

```python
import json

# JSON-Datei öffnen
with open("sessions.json", "r", encoding="utf-8") as f:
    daten = json.load(f) #json data gets defined as a list

# Einzelne Einträge lesen
for session in daten:
    print(f"{session['user_id']}: {session['duration']}s, Erfolg: {session['success']}") # session Position in Data json 0 > user_id, 1 > duration etc.

anzahl = len(daten)
print(f"\nAnzahl Sessions: {anzahl}")

# Gesamtdauer berechnen
gesamt_dauer = 0
for s in daten:
    dauer = s.get("duration", 0)
    gesamt_dauer += dauer

# Durchschnitt berechnen
if anzahl > 0:
    durchschnitt = gesamt_dauer / anzahl
else:
    durchschnitt = 0

print(f"Durchschnittliche Dauer: {durchschnitt:.1f}s")
```

### CSV-Datei lesen

```python
import csv

# CSV-Datei öffnen und lesen
with open("sessions.csv", "r", encoding="utf-8") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(f"User {row['user_id']}: {row['duration']}s, Erfolg: {row['success']}")
```

## Schreiben einer Datei
Dateien können mit `open()` im Schreibmodus (`'w'`) erstellt oder überschrieben werden.  
Mit `'a'` (append) kannst du neue Inhalte an eine bestehende Datei anhängen, ohne sie zu löschen.

### Textdatei schreiben
Im einfachsten Fall schreiben wir Textzeilen in eine `.txt`-Datei.

```python
# Textdatei schreiben
with open("ausgabe.txt", "w", encoding="utf-8") as file:
    file.write("Erste Zeile\n")
    file.write("Zweite Zeile\n")

# An Datei anhängen
with open("ausgabe.txt", "a", encoding="utf-8") as file:
    file.write("Dritte Zeile (angehängt)\n")

# Dateiinhalt zur Kontrolle lesen
with open("ausgabe.txt", "r", encoding="utf-8") as file:
    print(file.read())
```

### CSV-Datei schreiben
Mit dem Modul `csv` lassen sich strukturierte Daten zeilenweise schreiben.

```python
import csv

daten = [
    {"user_id": 1, "duration": 42, "success": True},
    {"user_id": 2, "duration": 58, "success": False},
]

# CSV-Datei schreiben
with open("sessions.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.DictWriter(f, fieldnames=["user_id", "duration", "success"])
    writer.writeheader()       # Kopfzeile
    writer.writerows(daten)    # Datenzeilen

# Kontrolle
with open("sessions.csv", "r", encoding="utf-8") as f:
    print(f.read())
```

### JSON-Datei schreiben
Das Schreiben von JSON-Dateien erfolgt mit `json.dump()`. Damit lassen sich komplexe Datenstrukturen (z. B. Listen oder Dictionaries) speichern.

```python
import json

daten = [
    {"user_id": 1, "duration": 42, "success": True},
    {"user_id": 2, "duration": 58, "success": False},
]

# JSON-Datei schreiben
with open("sessions.json", "w", encoding="utf-8") as f:
    json.dump(daten, f, indent=4, ensure_ascii=False)

# Kontrolle
with open("sessions.json", "r", encoding="utf-8") as f:
    print(f.read())
```

```python
# Schreiben: Erstellt eine neue Datei oder überschreibt eine bestehende
with open("daten.csv", "w") as file:
    file.write("user_id,duration,success,backtracks\n")
    file.write("u001,120,True,2\n")
    file.write("u002,95,False,4\n")

# Anhängen: Fügt neue Zeilen an eine bestehende Datei an
with open("daten.csv", "a") as file:
    file.write("u003,80,True,1\n")
```
